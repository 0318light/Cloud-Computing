```python
# 타입 힌트
# 파이썬에서 변수나 함수의 매개변수, 반환값이 어떤 타입이어야 하는지 코드에 명시적으로 표시하는 문법
# 파이썬 자체는 타입 힌트를 강제하지 않음 → 타입 힌트와 다른 타입을 사용해도 에러가 발생하지 않는다는 의미

# 반면에 FastAPI는 Pydantic을 이용해서 강제성을 적용하는 것
# 파이썬 인터프리터는 변수의 타입이 코드 작성 시점이 아니라, 실제 값이 할당되는 실행 시점에 결정되는 동적 타이핑 언어
# FastAPI는 여기서 한 걸음 더 나아가 Pydantic (BaseModel)의 도움을 받아 타입 체킹까지 하겠다는 것
# 타입 체킹을 해서 미리 정해진 타입과 다른 값이 들어오면 에러를 발생하겠다는 것
# 즉, 서버 측에서 타입 체킹을 엄격하게 하겠다는 것
# 타입 체킹을 안하면 발생하는 문제 -- 실행 에러, 보안 문제, 저장하고 보니 타입이 섞여 있는 데이터베이스

# 변수명: 타입
# 의 형태

# -> 타입
# 함수 리턴의 타입 힌트
```

```python
# 타입 분류

# 단순 타입 — str, int, bool, float
# 제네릭(컨테이너) 타입 — list[X], dict[K, V]
# BaseModel(중첩 모델) 타입 — User, list[Todo]
# Union 타입 — X | Y, Optional[X]
# 정해진 값만 허용하는 타입 — Enum, Literal
```

```python
# 단순 타입 - 1

# 기본 스칼라 타입
# str       - 문자열
# int       - 정수
# float     - 실수
# bool      - 참/거짓
# bytes     - 바이트 sequence → 0-255 사이의 값으로 이루어진 sequence

data1 = "안녕".encode("utf-8")      # UTF-8로 인코딩한 바이트
data2 = "안녕".encode("euc-kr")     # EUC-KR로 인코딩한 바이트

print(type(data1), type(data2))     # 인코딩이 달라도 타입은 bytes
print(data1)                        # 타입은 같지만 인코딩이 다르므로 내용물은 다름
print(data2)                        # 타입은 같지만 인코딩이 다르므로 내용물은 다름
                                    # bytes 값은 어떤 인코딩으로 만들어졌는지 알아야 원본을 복원할 수 있음

print(b"\x00\x01\x02")              # 알려져 있는(고정된) 바이트 값을 코드 내에서 사용하고자 하는 경우 b""의 형태로 사용
```

```python
# 단순 타입 - 2

# 날짜/시간 관련
# date        - 날짜 (연-월-일)
# time        - 시간
# datetime    - 날짜+시간
# timedelta   - 시간 간격

from datetime import date, time, datetime, timedelta

event_date: date = date(2026, 7, 22)                    # 날짜 (연-월-일)
start_time: time = time(14, 30, 0)                      # 시간
created_at: datetime = datetime(2026, 7, 22, 9, 0, 0)   # 날짜+시간
duration: timedelta = timedelta(hours=2, minutes=30)    # 시간 간격

print(type(event_date), event_date)
print(type(start_time), start_time)
print(type(created_at), created_at)
print(type(duration), duration)

# timedelta는 날짜/시간에 더하거나 뺄 수 있음
end_time: datetime = created_at + duration  # 09:00:00 + 02:30:00
print("end_time", type(end_time), end_time)
```

```python
# 단순 타입 - 3

# UUID      - 고유 식별자
# Decimal   - 정밀한 소수 (금액 계산 등)
# Path      - 파일 경로

from uuid import UUID, uuid4
from decimal import Decimal
from pathlib import Path

# UUID - 고유 식별자
user_id: UUID = uuid4()     # user_id 변수는 UUID 타입, UUID는 "Universally Unique Identifier"
                            # uuid4()는 123e4567-e89b-12d3-a456-426614174000와 같은 값을 생성
                            # 고유 식별자로 사용
print(type(user_id), user_id)

```

```python
# Decimal - 정확한 계산
# float 타입은 이진수로 실수를 표현하기 때문에 오차가 있을 수 있음
# float은 내부적으로 이진 분수(binary fraction)로 저장
# Decimal은 내부적으로 십진 분수(decimal fraction, base-10)로 저장

price: Decimal = Decimal("19.99")
tax: Decimal = Decimal("0.10")
total: Decimal = price + (price * tax)
print(type(total), "price=", total)

# float 계산(비교용)
price_float = 19.99
print(price_float + price_float * 0.10)
```

```python
# 특수한 타입

# None은 타입도 가능하고 값도 가능
# author: User | None = None
# 변수 author의 타입은 User 또는 None, 값은 None을 할당
# 어떤 변수의 타입에 None이 있어야 값으로 None을 할당 가능

# 타입 힌트가 없다면 None 할당 가능 (함수 파라미터도 동일하게 적용)
# x = None                  # 타입 힌트 없음 → 자유롭게 None 할당 가능
# x: str | None = None      # 타입 힌트에 None 포함 → 정상

# Any: 어떤 타입도 가능
from typing import Any

value: Any = "문자열"
value: Any = 123
value: Any = [1, 2, 3]
value: Any = {"key": "value"}
```

```python
# Generic을 타입 힌트로 어떻게 표현

# 타입의 성격을 말하는 것이지 타입 자체는 아님

# 리스트의 경우, 요소의 타입을 정할 수 있음 ("정수 리스트", "문자열 리스트")
# "정수 리스트"에서 정수가 리스트 타입의 매개 변수에 해당, "문자열 리스트"에서 문자열이 리스트 타입의 매개 변수에 해당
# 이와 같이 타입의 매개변수를 알려줘야 하는 타입을 지네릭이라고 함

# K, X, Y, Z는 모두 어떤 타입에 해당
# list[X]                   # 리스트 요소의 타입은 X
# dict[K, V]                # key의 타입 K, value의 타입 V
# tuple[X, Y]               # 타입이 X, Y인 요소가 정확히 2개인 튜플
# tuple[X, X]               # 타입이 X인 요소가 정확히 두 개인 튜플
# tuple[X, Y, Z]            # 타입이 X, Y, Z인 요소가 정확히 3개인 튜플

# tuple[X, ...]             # X 타입의 요소가 0개 이상, 개수 제한 없이
# 예를 들어 tuple[int, ...]의 경우
# ()                    # 0개
# (1,)                  # 1개
# (1, 2)                # 2개
# (1, 2, 3, 4, 5)       # 5개
# 6개, 7개, 8개, ...

# tuple[X, Y, ...]는 에러, 생략 기호는 반드시 타입 하나에만 사용


# set[X]                    # 타입이 X인 값들의 집합

def get_total_price(prices: list[float]) -> float:
    return sum(prices)


def merge_scores(scores: dict[str, int]) -> int:
    return sum(scores.values())


def get_coordinates() -> tuple[int, int]:
    return (3, 5)


def how_many_tags(tags: set[str]) -> int:
    return len(tags)

print(get_total_price([1000.0, 2500.0, 3200.0]))
print(merge_scores({"철수": 90, "영희": 85}))
print(get_coordinates())
print(how_many_tags({"급함", "중요", "급함"}))
```

```python
# Enum과 Literal

# 둘 다 "미리 정해둔 몇 개의 값"을 정의하는 용도로 사용
# 차이는 Enum은 별도의 클래스로 값들을 정의하는데 반해
# Literal은 클래스 없이 사용

from enum import Enum
from typing import Literal

class Priority(str, Enum): 
    # str, Enum을 상속
    # Enum을 상속: LOW, MEDIUM, HIGH처럼 미리 정해둔 값들만 가질 수 있는 열거형(Enum) 클래스가 됨
    # str도 상속: 멤버 Priority.LOW, Priority.MEDIUM, Priority.HIGH가 문자열처럼 동작 (예: Priority.LOW == "low"가 True)
    # Enum만 상속했다면 문자열과 직접 비교되지 않고 Priority.LOW라는 열거형 멤버로만 사용해야 함
    LOW = "low"             # LOW는 Enum의 멤버 이름, "low"는 해당 멤버의 실제 값
    MEDIUM = "medium"
    HIGH = "high"

# Enum을 타입 힌트에 적용
priority: Priority = Priority.HIGH  # 값 "low"는 멤버 HIGH를 통해서
# priority 변수의 타입은 Priority
# priority 변수의 값은 Priority.HIGH
print(priority, type(priority))

status: Literal["todo", "in_progress", "done"] = "in_progress"
# status 변수의 타입은 Literal["todo", "in_progress", "done"]
# status 값은 "in_progress"로 초기화

print(status, type(status))
```

```python
# Pydantic은 파이썬 데이터 검증/변환 라이브러리
# Pydantic = Python + Pedantic ("giving too much attention to formal rules or small details")
# BaseModel: Pydantic 라이브러리의 클래스 중 하나
# BaseModel은 FastAPI 뿐만 아니라 AI의 응답이 미리 정의한 응답 구조를 따르는지 검증하는 용도로도 사용
```

```python
from pydantic import BaseModel

# 어떤 클래스로 객체를 만들려고 하는데, 각 속성의 값이 어떤 기준이나 조건을 만족하는지가 중요하다
# 그 경우 BaseModel을 상속받으면 속성에 대해서 검증 기능을 사용할 수 있게 된다
# 검증 기준은 바로 속성에 적어놓은 타입 힌트이다

# 필드: 필수 필드와 선택 필드로 분류
# 필드 선언 시에 기본값이 있으면 선택적 필드, 기본값이 없으면 필수 필드
# "이 필드에 꼭 값을 넣는지 안 넣는지 지켜보겠음"
class TodoCreate(BaseModel):
    title: str                      # 필수 필드 (기본값 없음)
    completed: bool = False         # 선택적 필드 (기본값 False)

# 필수든 선택이든 채우면 문제 없음
TodoCreate(title="우유 사기", completed=True)

# 선택 필드는 값을 전달하지 않아도 문제 없음
TodoCreate(title="우유 사기")

# 필수는 채우지 않으면 문제
# TodoCreate(completed=True)

# 필드 이름이 다르면 전달하지 않은 것으로 되어 문제
# TodoCreate(name="우유 사기", completed=True)

# 필드 이름이 다르지만 선택적 필드는 값을 넘겨 주지 않아도 문제 없음
TodoCreate(title="우유 사기", finished=True)
```

```python
# Pydantic의 타입 변환
# 어떤 필드를 A 타입으로 선언했는데, 실제로 들어온 값이 B 타입일 때, Pydantic이 그 값을 A로 변환해서 받아줄 수 있는가?

# bool로 선언
# "true", "True", "1", 1, "yes", "on", 't', 'y' → True
# "false", "False", "0", 0, "no", "off", 'f', 'n' → False

# int로 선언
# "123" (숫자로만 이루어진 문자열) → 123
# 123.0 (소수점 이하 0인 float) → 123
# 123.5 → 에러 (정수로 변환 불가)

# float으로 선언
# "123.45" → 123.45
# 123 (int) → 123.0

# str으로 선언
# 오로지 str만 허용

# datetime으로 선언
# "2026-07-20", "2026-07-20T10:00:00" 같은 ISO 형식 문자열 → datetime
# 정수/실수(Unix timestamp) → datetime
# Unix timestamp가 실수인 경우, 소수점 이하 자릿수가 1자리면 1/10초 단위, 2자리면 1/100초 단위, ...


# 필드 선언이 list, dict 등 컨테이너 타입으로 선언된 경우

# 1. 리스트/딕셔너리 타입으로 값을 전달하면 그대로 사용, 문제 없음
# 2. 생긴 건 리스트/딕셔너리처럼 보이는데 알고 보니 JSON 문자열인 경우

# JSON 문자열 자동 파싱은 model_validate_json()으로 전체를 JSON 문자열로 줄 때만 적용됨
# (일반 생성자 방식 TypeConversionDemo(tags='[...]')으로는 파싱되지 않고 에러 발생)
```

```python
from pydantic import BaseModel
from datetime import datetime


class TypeConversionDemo(BaseModel):
    is_done: bool
    count: int
    price: float
    label: str
    created_at: datetime
    tags: list[str]


# bool 변환
result1 = TypeConversionDemo(
    is_done="true",       # "true" → True 로 변환되는지 확인
    count=1,
    price=1.0,
    label="x",
    created_at="2026-07-20",
    tags=["a"],
)
print(result1.is_done, type(result1.is_done))
```

```python
# int 변환
result2 = TypeConversionDemo(
    is_done=True,
    count="123",           # "123" → 123 으로 변환되는지 확인
    price=1.0,
    label="x",
    created_at="2026-07-20",
    tags=["a"],
)
print(result2.count, type(result2.count))
```

```python
# int 변환 에러 - 소수점 이하가 0이 아니면 변환 불가
# TypeConversionDemo(
#     is_done=True,
#     count=123.5,
#     price=1.0,
#     label="x",
#     created_at="2026-07-20",
#     tags=["a"],
# )
```

```python
# float 변환
result3 = TypeConversionDemo(
    is_done=True,
    count=1,
    price="123.45",        # "123.45" → 123.45 로 변환되는지 확인
    label="x",
    created_at="2026-07-20",
    tags=["a"],
)
print(result3.price, type(result3.price))
```

```python
# datetime 변환 - ISO 문자열
result4 = TypeConversionDemo(
    is_done=True,
    count=1,
    price=1.0,
    label="x",
    created_at="2026-07-20T10:00:00",   # 문자열 → datetime 객체로 변환되는지 확인
    tags=["a"],
)
print(result4.created_at, type(result4.created_at))
```

```python
# datetime 변환 - Unix timestamp(정수/실수)
result5 = TypeConversionDemo(
    is_done=True,
    count=1,
    price=1.0,
    label="x",
    created_at=1753000000.5,     # 숫자 → datetime 객체로 변환되는지 확인
    tags=["a"],
)
print(result5.created_at, type(result5.created_at))
```

```python
# BaseModel의 타입 힌트만으로는 부족한 세부적인 제약 조건을 각 속성에 적용하기 위한 목적으로 사용
# 예를 들어
# title: str → "문자열이어야 한다"는 타입 힌트로 가능
# 하지만 "문자열의 길이가 1자 이상 20자 이하여야 한다"는 조건은 타입 힌트만으로는 표현할 수 없음
# 그런 경우 Field()를 사용
# Field()는 단독으로 사용하는 것이 아니라 BaseModel 상속 클래스 내부, pydantic의 @dataclass와 함께 사용

from pydantic import BaseModel, Field

class TodoCreate(BaseModel):
    # 문자열 길이 제한
    title: str = Field(min_length=1, max_length=20)

    # 숫자 범위 제한 (gt: 초과, ge: 이상, lt: 미만, le: 이하)
    priority: int = Field(ge=1, le=5)

    # 정규식 패턴 - 영문 대소문자와 숫자로만 이루어진 코드
    code: str = Field(pattern=r"^[A-Za-z0-9]+$")
    # ^ 문자열의 시작
    # [A-Za-z0-9] 대문자/소문자/숫자
    # []+ 한 번 이상 반복
    # $ 문자열의 끝
    # 문자열 전체(처음부터 끝까지)가 영문 대소문자와 숫자로만 이루어져 있는 패턴

    # Field()로 기본값 지정 + 설명 (메타 정보) 추가
    completed: bool = Field(default=False, description="할 일 완료 여부")

# 정상 - 모든 제약 조건을 만족
result = TodoCreate(title="우유 사기", priority=3, code="ABC123")
print(result)
```

```python
# 커스텀 검증(Validator)

# Field()는 미리 정해진 규칙을 사용 → 정해진 규칙만 사용해서 검증이 가능
# Validator는 검증 로직을 직접 함수로 작성
# 아래의 형식으로 사용

# @field_validator("■") 또는 @model_validator        <--------- ■는 검증하려는 대상 속성의 이름
# @classmethod
# def 검증_함수_이름(cls, v):                         <--------- v는 ■에 전달된 실제 값
#     # v를 어떻게 검증할지, 필요하면 어떻게 변환할지에 대한 로직
#     if 조건_불만족:
#         raise ValueError("에러 메시지")
#     return v   # 통과하면 (변환된) 값을 반환
```

```python
# @field_validator — 단일 필드 검증

from pydantic import BaseModel, field_validator


class TodoCreate(BaseModel):
    title: str

    @field_validator("title")   # 이 함수는 필드 "title"에 대해서 검증
    @classmethod                # Pydantic의 validator 함수는 클래스 메서드 형태로 사용하는 경우
                                # 왜 클래스 함수?
                                # 검증은 인스턴스가 만들어지기 전에, 입력 데이터를 확인하는 단계에서 실행
                                # TodoCreate(title="...")에서 인스턴스를 만들기 전에 검증
                                # 이 단계는 인스턴스가 없으므로 클래스 함수 형태로 사용
    def no_bad_words(cls, v):   # cls: 클래스 TodoCreate, v: 검증 대상이 되는 실제 값 (title에 전달된 값)
        if "금지어" in v:
            raise ValueError("제목에 금지어를 포함할 수 없습니다")
            # raise ValueError(X)는 
            # 1. 검증 실패 예외를 발생시키면서
            # 2. 실패 이유를 문자열 X로 전달
        return v # return 값은 내부 검증 엔진이 받아가서 만들고자 하는 인스턴스의 "title" 속성에 전달


# 정상 - 금지어가 없으므로 통과
result = TodoCreate(title="우유 사기")
print(result)

# 에러 - "금지어"라는 단어가 포함되어 raise ValueError("제목에 금지어를 포함할 수 없습니다")
TodoCreate(title="이건 금지어가 들어간 제목")

# 예외 처리는 아래와 같이
# try:
#     TodoCreate(title="이건 금지어가 들어간 제목")
# except ValueError as e:
#     print("검증 실패:", e)
```

```python
# @field_validator — 하나의 validator를 여러 필드에 재 사용

from pydantic import BaseModel, field_validator

class TodoCreate(BaseModel):
    title: str
    description: str

    # 여러 필드 이름을 콤마로 나열하면, 하나의 validator를 여러 필드에 적용
    @field_validator("title", "description")
    @classmethod
    def no_bad_words(cls, v):
        # v
        # @field_validator("title", "description")에서 대상이 title, description 이므로
        # no_bad_words이 두 번 호출됨; 한 번은 v가 "title" 필드의 값, 또 한 번은 "description" 필드의 값으로
        if "금지어" in v:
            raise ValueError("금지어를 포함할 수 없습니다")
        return v


# 정상 - 둘 다 "금지어" 없음
result = TodoCreate(title="우유 사기", description="마트에서 구매")
print(result)

# 예외 발생 - title에 금지어 포함
# TodoCreate(title="이건 금지어가 들어간 제목", description="마트에서 구매")

# 예외 발생 - description에 금지어 포함
# TodoCreate(title="우유 사기", description="이것도 금지어가 있음")
```

```python
# @field_validator vs. @model_validator
# 무엇을 받아서 검증하는지가 다름

# @field_validator: 지정한 필드 하나의 값
# @model_validator: 모델에 전달된 모든 필드 값을 담은 데이터 전체

# 왜 @model_validator?
# 예를 들어 "시작일이 종료일보다 늦으면 안 된다"처럼 여러 필드를 동시에 고려 해야 하는 검증은 
# @field_validator로는 안되고, @model_validator로 전체 데이터를 받아서 처리해야 함
```

```python
# @model_validator - 여러 필드를 함께 사용해서 검증

# model_validator는 두 가지가 있음, "before validator"와 "after validator"
# 언제 validate를 할 것인지를 정하는 것, 인스턴스가 만들어지기 전? 후?

# @model_validator()를 사용함에 있어 mode는 디폴트 값이 "after"
# "before" 또는 "after" 중 사용

# (mode="before") → "before validator": are run before the model is instantiated
# (mode= "after") →  "after validator": run after the whole model has been validated

from pydantic import BaseModel, model_validator
from datetime import date

class EventAfter(BaseModel):
    start_date: date
    end_date: date

    @model_validator(mode="after")
    def check_dates(self):
        print("지금 이 시점은 인스턴스가 만들어진 시점")
        print(type(self.start_date), type(self.end_date))
        if self.start_date > self.end_date:
            raise ValueError("시작일이 종료일보다 늦을 수 없습니다")
        return self

EventAfter(start_date="2026-07-20", end_date="2026-07-25")

# BaseModel을 상속했기 때문에 각 필드(start_date, end_date)의 타입 검증/변환이 진행 (문자열 → date 객체로 변환 가능?)
# 모든 필드 검증/변환이 끝나고 인스턴스가 만들어짐

# 다음으로 mode="after"인 check_dates()가 이미 만들어진 인스턴스(self)에 대해서 검증
# self.start_date > self.end_date에 해당하지 않으므로 문제 없음

# EventAfter(start_date="2026-07-25", end_date="2026-07-20")
# 의 경우 start_date가 end_date보다 늦으므로 raise ValueError()
```

```python
from pydantic import BaseModel, ValidationError, model_validator

class UserModel(BaseModel):
    username: str
    password1: str
    password2: str

    @model_validator(mode='before') 
    # mode='before'인 경우, 인스턴스가 만들어지기 전 시점(self가 존재하지 않음) → @classmethod로 정의
    @classmethod
    def check_card_number_omitted(cls, data):
        # model validator이므로 전달받은 data는
        # 생성자에 넘긴 전체 데이터
        # 생성자를 UserModel(username="철수", password1="1234", password2="1234")
        # 와 같이 호출했다면 data는 딕셔너리: {"username": "철수", "password1": "1234", "password2": "1234"}
        
        # 원칙적으로 isinstance(data, dict)로 확인한 뒤 처리해야 함 → data가 딕셔너리라는 보장이 없기 때문
        assert 'card_number' not in data, 'card_number should not be included'
        # assert A, B의 형태
        # → A가 참(True)이 아니면, 메시지 B를 사용해서 에러(AssertionError)를 발생
        
        # 넘어온 data의 키 중에 'card_number'가 있으면 안된다는 상황
        
        return data

    @model_validator(mode='after')
    def check_passwords_match(self):
        if self.password1 != self.password2:
            raise ValueError('passwords do not match')
        return self
    
# UserModel(username="철수", password1="1234", password2="1235")
# before 단계에서 실패 → 인스턴스 만들어지지 않음
#  after 단계에서 실패 → 최종적으로 인스턴스가 반환되지 않음

UserModel(username="철수", password1="1234", password2="1234")
```

```python
# 지금까지는
# (BaseModel을 상속받은) 클래스로 인스턴스를 만드는 과정에서 검증 및 변환을 수행했음

# 인스턴스를 만드는 형태는
# 1.
# UserModel(username="철수", password1="1234", password2="1234")
# 2.
# data = {"username": "철수", "password1": "1234", "password2": "1234"}
# UserModel(**data)

# model_config는 BaseModel에 정의된 속성 → BaseModel을 상속받으면 사용 가능
# Model_config는 실질적으로 딕셔너리 타입 (ConfigDict)
# from_attributes는 Model_config의 키이며 값은 True/False

# from_attributes: True
# 위의 1, 2 방법 뿐만 아니라, 이미 존재하는 일반 객체의 속성(attribute)에서도 값을 읽어와서 인스턴스를 만들 수 있음
```

```python
from pydantic import BaseModel

class TodoOut(BaseModel):
    id: int
    title: str

TodoOut(id=1, title="우유 사기") # 일반적인 경우
# -------------------------------------------
class SomeObject:
    def __init__(self, id, title):
        self.id = id
        self.title = title

obj = SomeObject(id=1, title="우유 사기") # 일반 파이썬 객체 obj
# -------------------------------------------
class TodoOutFromAttr(BaseModel):
    model_config = {"from_attributes": True}
    # BaseModel + 일반 객체 obj = SomeObject(id=1, title="우유 사기")의 속성에서 값을 가져와서 사용
    id: int
    title: str

result = TodoOutFromAttr.model_validate(obj)
# 일반 객체 obj의 속성 obj.id, obj.title을
# TodoOutFromAttr에 정의된 필드 타입(id: int, title: str)에 맞게 검증/변환한 뒤
# TodoOutFromAttr의 인스턴스를 만들어 result로 반환
print(result)
```

```python
# 딕셔너리를 전달
data_missing = {"id": 1}   # title이 없음
TodoOutFromAttr.model_validate(data_missing)
# TodoOutFromAttr에 필수 필드 title이 있는데 값을 전달하지 않음
```

```python
# 딕셔너리에는 있는데, 모델에 없는 필드
data_extra = {"id": 1, "title": "우유 사기", "extra_field": "추가된 값"}
result_extra = TodoOutFromAttr.model_validate(data_extra)
print(result_extra) # 남아도는 것은 문제 없음
```

```python
# 일반 객체에는 있는데, 모델에 없는 필드
class ObjectExtraField:
    def __init__(self, id, title, extra_field):
        self.id = id
        self.title = title
        self.extra_field = extra_field   # 모델에 없는 속성

obj_extra = ObjectExtraField(id=1, title="우유 사기", extra_field="추가된 값") # 일반 객체
result_extra_attr = TodoOutFromAttr.model_validate(obj_extra)
print(result_extra_attr) # 역시 남아도는 것은 문제 없음
```
